## 5. Sampling & Distribution Assumptions

Weighted sampling is used to create a more realistic distribution of travel bookings across countries, destination types, trip purposes, accommodation categories, transportation modes, and seasons.

These weights are synthetic assumptions created for the purpose of generating a realistic analytical dataset.

### 5.1 Country Sampling Weights

```python
country_weights = {
    "France": 12,
    "Italy": 11,
    "Switzerland": 8,
    "Japan": 10,
    "Thailand": 13,
    "Singapore": 8,
    "UAE": 9,
    "Australia": 7,
    "Canada": 6,
    "USA": 10,
    "Spain": 9,
    "Greece": 6,
    "South Africa": 4,
    "Egypt": 5,
    "India": 12,
    "Brazil": 4,
    "Peru": 3,
    "Saudi Arabia": 4,
    "Qatar": 2,
    "New Zealand": 5
}
```

### 5.2 Destination Type Weights

```python
destination_weights = {
    "Beach": 28,
    "City": 22,
    "Mountain": 18,
    "Adventure": 14,
    "Heritage": 10,
    "Wildlife": 8
}
```

### 5.3 Travel Purpose Weights

```python
purpose_weights = {
    "Leisure": 45,
    "Business": 18,
    "Family Vacation": 18,
    "Adventure": 12,
    "Honeymoon": 7
}
```

### 5.4 Hotel Category Weights

```python
hotel_weights = {
    "3 Star": 35,
    "4 Star": 45,
    "5 Star": 20
}
```

### 5.5 Transportation Mode Weights

```python
transport_weights = {
    "Flight": 60,
    "Train": 15,
    "Car": 15,
    "Cruise": 10
}
```

### 5.6 Travel Season Weights

```python
season_weights = {
    "Spring": 18,
    "Summer": 28,
    "Monsoon": 12,
    "Autumn": 18,
    "Winter": 24
}
```

## 6. Helper Functions

Reusable helper functions are defined to standardize the generation of identifiers, demographic attributes, and hotel ratings.

Separating these rules into functions improves readability, maintainability, and consistency throughout the dataset-generation process.

### 6.1 Identifier & Demographic Generators

```python
# Booking ID
def generate_booking_id(index):
    """Generate a unique booking identifier."""
    return f"BK{index:06d}"

# Tourist ID
def generate_tourist_id():
    """Generate a synthetic tourist identifier."""
    return f"T{random.randint(100000,999999)}"

# Age
def generate_age():
    """Generate a synthetic traveler age between 18 and 70."""
    return int(np.clip(np.random.normal(35, 12), 18, 70))

# Hotel Rating
def generate_hotel_rating(category):

    if category=="5 Star":
        return round(random.uniform(4.5,5.0),1)

    elif category=="4 Star":
        return round(random.uniform(4.0,4.7),1)

    else:
        return round(random.uniform(3.5,4.3),1)
```

### 6.2 Stay Duration Generator

```python
def generate_stay_days(purpose):
    """Generate trip duration based on the travel purpose."""

    if purpose=="Business":
        return random.randint(2,5)

    elif purpose=="Honeymoon":
        return random.randint(5,10)

    elif purpose=="Family Vacation":
        return random.randint(4,8)

    elif purpose=="Adventure":
        return random.randint(4,9)

    else:
        return random.randint(3,7)
```

## 7. Cost Modeling Assumptions

Destination-level cost ranges are defined to simulate realistic variation in travel expenditure across different destination types.

These values are synthetic assumptions and are used to derive individual trip expense components.

### 7.1 Destination Cost Ranges

```python
destination_price = {

    "Beach": (30000,90000),

    "City": (25000,80000),

    "Mountain": (25000,85000),

    "Adventure": (45000,150000),

    "Heritage": (25000,70000),

    "Wildlife": (35000,120000)
}
```

## 8. Synthetic Travel Dataset Generation

The dataset is generated iteratively at the booking level.

Each record combines traveler characteristics, booking dates, destination information, transportation, accommodation, trip duration, expenditure components, travel purpose, seasonality, and satisfaction.

The generation logic uses the predefined sampling distributions and business rules established above.

```python
# Generate booking-level records
records = []

for i in range(1, num_records + 1):

 # 1. Booking & traveler information
    booking_id = generate_booking_id(i)
    tourist_id = generate_tourist_id()

    booking_date = fake.date_between(start_date="-2y", end_date="today")
    travel_date = booking_date + timedelta(days=random.randint(7,90))

    age = generate_age()

    gender = random.choice(["Male","Female"])

    nationality = random.choices(
        list(country_weights.keys()),
        weights=list(country_weights.values()),
        k=1
    )[0]

    travel_group = random.choices(
        travel_groups,
        weights=[25,30,30,15],
        k=1
    )[0]

    trip_type = random.choices(
        trip_types,
        weights=[30,70],
        k=1
    )[0]

    if trip_type == "Domestic":
        country = nationality

    else:

        available = [
            c for c in country_city.keys()
            if c != nationality
        ]

        weights = [
            country_weights[c]
            for c in available
        ]

        country = random.choices(
            available,
            weights=weights,
            k=1
        )[0]

    city = random.choice(country_city[country])

    continent = continent_map[country]

    # 2. Destination and geographic information

    destination_type = random.choices(
        list(destination_weights.keys()),
        weights=list(destination_weights.values()),
        k=1
    )[0]

    travel_purpose = random.choices(
        list(purpose_weights.keys()),
        weights=list(purpose_weights.values()),
        k=1
    )[0]

    hotel_category = random.choices(
        list(hotel_weights.keys()),
        weights=list(hotel_weights.values()),
        k=1
    )[0]

    transport_mode = random.choices(
        list(transport_weights.keys()),
        weights=list(transport_weights.values()),
        k=1
    )[0]

    travel_season = random.choices(
        list(season_weights.keys()),
        weights=list(season_weights.values()),
        k=1
    )[0]

    # 3. Travel characteristics

    stay_days = generate_stay_days(travel_purpose)

    return_date = travel_date + timedelta(days=stay_days)

    if transport_mode == "Flight":
        travel_duration = round(random.uniform(2,12),1)

    elif transport_mode == "Train":
        travel_duration = round(random.uniform(3,18),1)

    elif transport_mode == "Cruise":
        travel_duration = round(random.uniform(24,120),1)

    else:
        travel_duration = round(random.uniform(2,10),1)

    hotel_rating = generate_hotel_rating(hotel_category)

    # 4. Accommodation and stay information

    min_cost,max_cost = destination_price[destination_type]

    transport_cost = random.randint(
        int(min_cost*0.20),
        int(max_cost*0.35)
    )

    hotel_cost = random.randint(
        int(min_cost*0.30),
        int(max_cost*0.40)
    )

    food_cost = random.randint(
        int(min_cost*0.08),
        int(max_cost*0.15)
    )

    shopping_cost = random.randint(
        int(min_cost*0.05),
        int(max_cost*0.20)
    )

    local_transport_cost = random.randint(
        int(min_cost*0.03),
        int(max_cost*0.08)
    )

    activities_cost = random.randint(
        int(min_cost*0.08),
        int(max_cost*0.20)
    )

    # 5. Trip cost calculation
    total_trip_cost = (
        transport_cost
        + hotel_cost
        + food_cost
        + shopping_cost
        + local_transport_cost
        + activities_cost
    )

    # 6. Customer satisfaction

    satisfaction = round(
        np.clip(
            hotel_rating + np.random.normal(0,0.25),
            3.0,
            5.0
        ),
        1
    )

    # 7. Store completed booking record

    records.append([

        booking_id,
        booking_date,
        travel_date,
        return_date,

        tourist_id,

        age,
        gender,

        nationality,

        travel_group,
        trip_type,

        continent,
        country,
        city,

        destination_type,

        transport_mode,
        travel_duration,

        hotel_category,
        hotel_rating,

        stay_days,

        transport_cost,
        hotel_cost,
        food_cost,
        shopping_cost,
        local_transport_cost,
        activities_cost,

        travel_purpose,
        travel_season,

        satisfaction,

        total_trip_cost
    ])
```

## 9. DataFrame Construction

The generated booking records are converted into a Pandas DataFrame using an explicit column schema to ensure consistent field names and ordering.

```python
columns = [
    "Booking_ID",
    "Booking_Date",
    "Travel_Date",
    "Return_Date",
    "Tourist_ID",
    "Age",
    "Gender",
    "Nationality",
    "Travel_Group",
    "Trip_Type",
    "Continent",
    "Country",
    "City",
    "Destination_Type",
    "Transport_Mode",
    "Travel_Duration_Hours",
    "Hotel_Category",
    "Hotel_Rating",
    "Stay_Days",
    "Transport_Cost",
    "Hotel_Cost",
    "Food_Cost",
    "Shopping_Cost",
    "Local_Transport_Cost",
    "Activities_Cost",
    "Travel_Purpose",
    "Travel_Season",
    "Satisfaction_Rating",
    "Total_Trip_Cost"
]

df = pd.DataFrame(records, columns=columns)
```
